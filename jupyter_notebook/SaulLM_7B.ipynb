```python
# ========================================
# CELL 4: ENHANCED DOCUMENT PROCESSOR WITH OCR
# ========================================

class EnhancedDocumentProcessor:
    """Enhanced document processor with OCR capabilities"""

    def __init__(self):
        self.supported_formats = ['.pdf', '.txt', '.docx', '.png', '.jpg', '.jpeg', '.tiff', '.bmp']
        print("📄 Document processor with OCR ready!")

    def preprocess_image_for_ocr(self, image):
        """Preprocess image for better OCR results"""
        # Convert PIL Image to OpenCV format
        img_array = np.array(image)

        # Convert to grayscale
        if len(img_array.shape) == 3:
            gray = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)
        else:
            gray = img_array

        # Apply denoising
        denoised = cv2.fastNlMeansDenoising(gray)

        # Apply thresholding to get better contrast
        _, thresh = cv2.threshold(denoised, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

        return Image.fromarray(thresh)

    def extract_text_from_image(self, image_bytes: bytes, filename: str = "") -> str:
        """Extract text from image using OCR"""
        try:
            # Load image
            image = Image.open(BytesIO(image_bytes))

            # Preprocess for better OCR
            processed_image = self.preprocess_image_for_ocr(image)

            # Perform OCR with multiple configurations
            custom_config = r'--oem 3 --psm 6 -c tessedit_char_whitelist=ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789.,;:!?()[]{}"\'-/$%&*+=<>@#^_|\\~` '

            text = pytesseract.image_to_string(processed_image, config=custom_config)

            if not text.strip():
                # Fallback: try different PSM modes
                for psm in [3, 4, 6, 8, 11]:
                    config = f'--oem 3 --psm {psm}'
                    text = pytesseract.image_to_string(processed_image, config=config)
                    if text.strip():
                        break

            return text.strip() or "No text detected in image"

        except Exception as e:
            return f"Error extracting text from image: {str(e)}"

    def extract_text_from_pdf_with_ocr(self, file_bytes: bytes) -> str:
        """Extract text from PDF with OCR fallback for scanned documents"""
        text = ""

        try:
            # First, try normal text extraction
            pdf_reader = PyPDF2.PdfReader(BytesIO(file_bytes))

            for page_num, page in enumerate(pdf_reader.pages):
                try:
                    page_text = page.extract_text().strip()

                    # If page has little or no text, use OCR
                    if len(page_text) < 50:  # Threshold for "little text"
                        print(f"📸 Using OCR for page {page_num + 1} (scanned content detected)")

                        # Convert PDF page to image
                        images = convert_from_bytes(
                            file_bytes,
                            first_page=page_num + 1,
                            last_page=page_num + 1,
                            dpi=300
                        )

                        if images:
                            # Convert image to bytes for OCR
                            img_byte_arr = BytesIO()
                            images[0].save(img_byte_arr, format='PNG')
                            img_bytes = img_byte_arr.getvalue()

                            # Extract text using OCR
                            ocr_text = self.extract_text_from_image(img_bytes, f"page_{page_num + 1}")
                            text += f"\n--- Page {page_num + 1} (OCR) ---\n{ocr_text}\n"
                        else:
                            text += f"\n--- Page {page_num + 1} ---\n{page_text}\n"
                    else:
                        text += f"\n--- Page {page_num + 1} ---\n{page_text}\n"

                except Exception as e:
                    print(f"Warning: Could not process page {page_num + 1}: {e}")
                    continue

            return text.strip()

        except Exception as e:
            # If PDF reading fails completely, try full OCR
            print(f"📸 PDF text extraction failed, using full OCR: {e}")
            try:
                images = convert_from_bytes(file_bytes, dpi=300)
                full_text = ""

                for i, image in enumerate(images):
                    img_byte_arr = BytesIO()
                    image.save(img_byte_arr, format='PNG')
                    img_bytes = img_byte_arr.getvalue()

                    page_text = self.extract_text_from_image(img_bytes, f"page_{i + 1}")
                    full_text += f"\n--- Page {i + 1} (Full OCR) ---\n{page_text}\n"

                return full_text.strip()

            except Exception as ocr_error:
                return f"Error extracting PDF with OCR: {str(ocr_error)}"

    def extract_text_from_docx(self, file_bytes: bytes) -> str:
        """Extract text from DOCX file"""
        try:
            doc = docx.Document(BytesIO(file_bytes))
            text = ""

            # Extract paragraphs
            for paragraph in doc.paragraphs:
                if paragraph.text.strip():
                    text += paragraph.text + "\n"

            # Extract tables
            for table in doc.tables:
                for row in table.rows:
                    row_text = []
                    for cell in row.cells:
                        row_text.append(cell.text.strip())
                    if any(row_text):
                        text += " | ".join(row_text) + "\n"

            return text.strip()

        except Exception as e:
            return f"Error extracting DOCX: {str(e)}"

    def extract_text_from_txt(self, file_bytes: bytes) -> str:
        """Extract text from TXT file"""
        encodings = ['utf-8', 'latin-1', 'cp1252', 'iso-8859-1']

        for encoding in encodings:
            try:
                return file_bytes.decode(encoding)
            except UnicodeDecodeError:
                continue

        return "Error: Could not decode text file"

    def get_file_extension(self, filename: str) -> str:
        """Get file extension"""
        return filename.split('.')[-1].lower()

    def process_uploaded_file(self, file_bytes: bytes, filename: str) -> Dict:
        """Process uploaded file and return text with metadata"""
        ext = self.get_file_extension(filename)

        print(f"🔄 Processing {filename} ({ext.upper()} format)...")

        if ext == 'pdf':
            text = self.extract_text_from_pdf_with_ocr(file_bytes)
        elif ext == 'docx':
            text = self.extract_text_from_docx(file_bytes)
        elif ext == 'txt':
            text = self.extract_text_from_txt(file_bytes)
        elif ext in ['png', 'jpg', 'jpeg', 'tiff', 'bmp']:
            text = self.extract_text_from_image(file_bytes, filename)
        else:
            text = f"Unsupported file format: {ext}"

        return {
            "text": text,
            "filename": filename,
            "format": ext,
            "word_count": len(text.split()) if not text.startswith("Error") else 0,
            "char_count": len(text) if not text.startswith("Error") else 0,
            "processing_status": "success" if not text.startswith("Error") else "error"
        }
```

```python
# ========================================
# CELL 5: ENHANCED CONTRACT ANALYZER
# ========================================

class EnhancedContractAnalyzer:
    """Enhanced contract analyzer following ChainSight methodology"""

    def __init__(self):
        self.model = None
        self.tokenizer = None
        self.device = "cuda" if torch.cuda.is_available() else "cpu"
        self.knowledge_base = EnhancedIndustryKnowledge()
        self.document_processor = EnhancedDocumentProcessor()

        print(f"🔧 Enhanced Analyzer initialized - Device: {self.device}")

        # Enhanced risk keywords with weights
        self.risk_keywords = {
            "financial": {
                "keywords": ["payment", "penalty", "fee", "cost", "invoice", "liability", "damages", "compensation"],
                "weight": 1.2
            },
            "legal": {
                "keywords": ["liability", "lawsuit", "jurisdiction", "dispute", "breach", "violation", "court"],
                "weight": 1.5
            },
            "operational": {
                "keywords": ["performance", "delivery", "deadline", "sla", "service level", "availability", "uptime"],
                "weight": 1.0
            },
            "compliance": {
                "keywords": ["regulation", "standard", "certification", "compliance", "audit", "gdpr", "privacy"],
                "weight": 1.3
            },
            "intellectual_property": {
                "keywords": ["intellectual property", "copyright", "patent", "trademark", "trade secret", "proprietary"],
                "weight": 1.1
            },
            "termination": {
                "keywords": ["termination", "cancellation", "end", "expiry", "dissolution", "breach"],
                "weight": 1.0
            }
        }

    def load_model(self):
        """Load SaulLM-7B model with enhanced error handling"""
        print("🤖 Loading SaulLM-7B legal specialist model...")

        try:
            # Enhanced quantization config
            quantization_config = BitsAndBytesConfig(
                load_in_4bit=True,
                bnb_4bit_quant_type="nf4",
                bnb_4bit_compute_dtype=torch.float16,
                bnb_4bit_use_double_quant=True
            )

            model_name = "Equall/Saul-7B-Instruct-v1"

            # Load tokenizer
            self.tokenizer = AutoTokenizer.from_pretrained(model_name)
            if self.tokenizer.pad_token is None:
                self.tokenizer.pad_token = self.tokenizer.eos_token

            # Load model with enhanced settings
            self.model = AutoModelForCausalLM.from_pretrained(
                model_name,
                quantization_config=quantization_config,
                device_map="auto",
                trust_remote_code=True,
                torch_dtype=torch.float16
            )

            print("✅ SaulLM-7B loaded successfully!")
            return True

        except Exception as e:
            print(f"❌ Failed to load SaulLM-7B: {e}")
            print("💡 Trying fallback models...")

            # Try alternative legal models
            fallback_models = [
                "microsoft/DialoGPT-medium",
                "gpt2-medium"
            ]

            for fallback_model in fallback_models:
                try:
                    print(f"🔄 Attempting {fallback_model}...")
                    self.tokenizer = AutoTokenizer.from_pretrained(fallback_model)
                    if self.tokenizer.pad_token is None:
                        self.tokenizer.pad_token = self.tokenizer.eos_token

                    self.model = AutoModelForCausalLM.from_pretrained(fallback_model)
                    print(f"✅ Fallback model {fallback_model} loaded!")
                    return True
                except:
                    continue

            print("❌ Could not load any model")
            return False

    def analyze_contract(self, contract_text: str, industry: str = "technology") -> Dict:
        """Enhanced contract analysis with ChainSight methodology"""

        if not self.model:
            return {"error": "Model not loaded. Please run load_model() first."}

        print(f"🔍 Performing ChainSight analysis for {industry} industry...")

        # Get industry context
        industry_context = self.knowledge_base.get_industry_context(industry)

        # Basic metadata
        word_count = len(contract_text.split())
        char_count = len(contract_text)

        # Enhanced risk assessment
        risk_assessment = self._enhanced_risk_assessment(contract_text, industry_context)

        # Missing clauses analysis
        missing_clauses = self._analyze_missing_clauses(contract_text, industry_context)

        # AI-powered analysis
        ai_analysis = self._get_enhanced_ai_analysis(contract_text, industry, industry_context)

        # Generate recommendations using ChainSight methodology
        recommendations = self._generate_chainsight_recommendations(
            risk_assessment, missing_clauses, industry_context
        )

        # Calculate overall scores
        overall_risk = self._calculate_overall_risk(risk_assessment)
        completeness_score = self._calculate_completeness_score(missing_clauses, industry_context)

        return {
            "contract_info": {
                "word_count": word_count,
                "char_count": char_count,
                "industry": industry,
                "analysis_date": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                "analysis_method": "ChainSight Enhanced Analysis"
            },
            "executive_summary": {
                "overall_risk_score": overall_risk,
                "completeness_score": completeness_score,
                "critical_issues": len([r for r in missing_clauses if r.get("importance") == "critical"]),
                "priority_level": self._determine_priority_level(overall_risk, completeness_score)
            },
            "risk_assessment": risk_assessment,
            "missing_clauses": missing_clauses,
            "ai_analysis": ai_analysis,
            "recommendations": recommendations,
            "compliance_analysis": self._analyze_compliance(contract_text, industry_context)
        }

    def _enhanced_risk_assessment(self, text: str, industry_context: Dict) -> Dict:
        """Enhanced risk assessment with industry-specific weighting"""
        text_lower = text.lower()
        risk_scores = {}
        risk_details = {}

        for risk_type, config in self.risk_keywords.items():
            keywords = config["keywords"]
            weight = config["weight"]

            score = 0
            found_keywords = []

            for keyword in keywords:
                count = text_lower.count(keyword.lower())
                if count > 0:
                    score += count * weight
                    found_keywords.append(f"{keyword}({count})")

            # Normalize score to 1-10 scale with industry adjustment
            normalized_score = min(10, max(1, score))

            # Apply industry-specific adjustments
            industry_risks = industry_context.get("risk_factors", [])
            if any(risk_type in risk.lower() for risk in industry_risks):
                normalized_score *= 1.2  # Increase weight for industry-relevant risks

            risk_scores[risk_type] = round(min(10, normalized_score), 2)
            risk_details[risk_type] = {
                "score": risk_scores[risk_type],
                "found_keywords": found_keywords,
                "severity": self._get_severity(risk_scores[risk_type]),
                "industry_relevant": any(risk_type in risk.lower() for risk in industry_risks)
            }

        return {
            "risk_breakdown": risk_details,
            "high_risk_areas": [risk for risk, score in risk_scores.items() if score >= 7],
            "total_risks_identified": len([r for r in risk_scores.values() if r >= 4])
        }

    def _analyze_missing_clauses(self, text: str, industry_context: Dict) -> List[Dict]:
        """Analyze missing clauses with enhanced detection"""
        text_lower = text.lower()
        critical_clauses = industry_context.get("critical_clauses", [])

        # Enhanced clause detection patterns
        clause_patterns = {
            "Data Protection and Privacy": [
                "data protection", "privacy policy", "gdpr", "personal data",
                "data processing", "privacy compliance", "data security"
            ],
            "Intellectual Property Rights": [
                "intellectual property", "ip rights", "copyright", "patent",
                "trademark", "trade secret", "proprietary", "ownership"
            ],
            "Liability Limitations": [
                "liability", "limitation of liability", "damages", "indemnification",
                "hold harmless", "liability cap", "exclusion of damages"
            ],
            "Termination": [
                "termination", "end of agreement", "contract expiry", "cancellation",
                "dissolution", "breach termination", "notice period"
            ],
            "Service Level Agreements": [
                "sla", "service level", "performance standards", "uptime",
                "availability", "response time", "service quality"
            ],
            "Payment Terms": [
                "payment", "invoice", "billing", "fee schedule", "compensation",
                "payment terms", "due date", "late fees"
            ],
            "Confidentiality": [
                "confidential", "non-disclosure", "nda", "proprietary information",
                "confidentiality agreement", "trade secrets"
            ],
            "Force Majeure": [
                "force majeure", "act of god", "unforeseeable circumstances",
                "natural disaster", "pandemic", "government action"
            ],
            "Insurance Requirements": [
                "insurance", "coverage", "liability insurance", "professional liability",
                "general liability", "insurance certificate"
            ],
            "Dispute Resolution": [
                "dispute resolution", "arbitration", "mediation", "jurisdiction",
                "governing law", "legal proceedings", "court"
            ]
        }

        missing_clauses = []
        present_clauses = []

        for clause_name, patterns in clause_patterns.items():
            found = any(pattern in text_lower for pattern in patterns)
            found_patterns = [p for p in patterns if p in text_lower]

            if found:
                present_clauses.append({
                    "clause": clause_name,
                    "found_patterns": found_patterns,
                    "coverage": "present"
                })
            else:
                # Determine importance based on industry context
                is_critical = any(clause_name.lower() in critical.lower() for critical in critical_clauses)

                missing_clauses.append({
                    "clause": clause_name,
                    "importance": "critical" if is_critical else "recommended",
                    "reason": f"Essential for {industry_context.get('industry', 'general')} contracts" if is_critical else "Standard business practice",
                    "suggested_patterns": patterns[:3],  # Show top 3 patterns
                    "risk_level": "high" if is_critical else "medium"
                })

        return missing_clauses

    def _get_enhanced_ai_analysis(self, text: str, industry: str, industry_context: Dict) -> str:
        """Enhanced AI analysis with ChainSight methodology"""

        critical_clauses = industry_context.get("critical_clauses", [])
        risk_factors = industry_context.get("risk_factors", [])
        regulations = industry_context.get("regulations", [])

        # Enhanced prompt following ChainSight methodology
        prompt = f"""[INST] You are an expert ChainSight Contract AI Intelligence Analyst specializing in {industry.upper()} industry contracts.

CHAINSIGHT ANALYSIS FRAMEWORK:
1. ASSESSMENT: Evaluate contract capability across legal, financial, and operational dimensions
2. IMPLEMENTATION: Identify gaps and improvement opportunities
3. OPTIMIZATION: Provide strategic recommendations for contract enhancement

INDUSTRY CONTEXT - {industry.upper()}:
Critical Clauses: {', '.join(critical_clauses[:5])}
Key Risk Factors: {', '.join(risk_factors[:5])}
Relevant Regulations: {', '.join(regulations[:3])}

CONTRACT TO ANALYZE:
{text[:1500]}...

PROVIDE CHAINSIGHT ANALYSIS:
1. Contract Assessment Score (1-10)
2. Key Legal Risks Identified
3. Financial Protection Adequacy
4. Operational Risk Factors
5. Strategic Recommendations
6. Compliance Gaps
7. Implementation Priority

Be specific, actionable, and industry-focused: [/INST]"""

        try:
            inputs = self.tokenizer(prompt, return_tensors="pt", max_length=2000, truncation=True)

            with torch.no_grad():
                outputs = self.model.generate(
                    inputs.input_ids,
                    max_new_tokens=300,
                    temperature=0.3,  # Lower temperature for more focused analysis
                    do_sample=True,
                    pad_token_id=self.tokenizer.eos_token_id,
                    repetition_penalty=1.1
                )

            response = self.tokenizer.decode(outputs[0], skip_special_tokens=True)

            # Extract generated content
            if "[/INST]" in response:
                analysis = response.split("[/INST]")[-1].strip()
            else:
                analysis = response

            return analysis

        except Exception as e:
            return f"ChainSight AI analysis unavailable: {str(e)}"

    def _generate_chainsight_recommendations(self, risk_assessment: Dict, missing_clauses: List, industry_context: Dict) -> List[Dict]:
        """Generate recommendations using ChainSight methodology"""
        recommendations = []

        # Assessment phase recommendations
        high_risks = risk_assessment.get("high_risk_areas", [])
        if high_risks:
            recommendations.append({
                "phase": "Assessment",
                "priority": "HIGH",
                "category": "Risk Mitigation",
                "description": f"Address critical risk areas: {', '.join(high_risks)}",
                "action": "Implement immediate risk controls and monitoring",
                "timeline": "1-2 weeks"
            })

        # Implementation phase recommendations
        critical_missing = [c for c in missing_clauses if c.get("importance") == "critical"]
        if critical_missing:
            recommendations.append({
                "phase": "Implementation",
                "priority": "HIGH",
                "category": "Contract Enhancement",
                "description": f"Add {len(critical_missing)} critical missing clauses",
                "action": f"Include: {', '.join([c['clause'] for c in critical_missing[:3]])}",
                "timeline": "1-3 weeks"
            })

        # Optimization phase recommendations
        total_risks = risk_assessment.get("total_risks_identified", 0)
        if total_risks >= 3:
            recommendations.append({
                "phase": "Optimization",
                "priority": "MEDIUM",
                "category": "Process Improvement",
                "description": "Establish ongoing contract monitoring and review process",
                "action": "Implement quarterly contract health checks",
                "timeline": "1-2 months"
            })

        return recommendations

    def _get_severity(self, score: float) -> str:
        """Get risk severity level"""
        if score >= 7:
            return "High"
        elif score >= 4:
            return "Medium"
        else:
            return "Low"

    def _calculate_overall_risk(self, risk_assessment: Dict) -> float:
        """Calculate overall risk score"""
        risk_breakdown = risk_assessment.get("risk_breakdown", {})
        if not risk_breakdown:
            return 0.0

        scores = [details["score"] for details in risk_breakdown.values()]
        return round(np.mean(scores), 2)

    def _calculate_completeness_score(self, missing_clauses: List, industry_context: Dict) -> float:
        """Calculate contract completeness score"""
        total_critical = len(industry_context.get("critical_clauses", []))
        missing_critical = len([c for c in missing_clauses if c.get("importance") == "critical"])

        if total_critical == 0:
            return 10.0

        completeness = ((total_critical - missing_critical) / total_critical) * 10
        return round(max(0.0, completeness), 2)

    def _determine_priority_level(self, risk_score: float, completeness_score: float) -> str:
        """Determine overall priority level"""
        if risk_score >= 7 or completeness_score <= 5:
            return "HIGH"
        elif risk_score >= 4 or completeness_score <= 7:
            return "MEDIUM"
        else:
            return "LOW"

    def _analyze_compliance(self, text: str, industry_context: Dict) -> Dict:
        """Analyze regulatory compliance"""
        text_lower = text.lower()
        regulations = industry_context.get("regulations", [])

        compliance_status = {}
        for regulation in regulations:
            # Check for regulation mentions
            reg_keywords = {
                "GDPR": ["gdpr", "general data protection", "data protection regulation"],
                "CCPA": ["ccpa", "california consumer privacy", "consumer privacy act"],
                "SOX": ["sarbanes-oxley", "sox", "financial reporting"],
                "HIPAA": ["hipaa", "health insurance portability", "medical privacy"],
                "OSHA": ["osha", "occupational safety", "workplace safety"],
                "EPA": ["epa", "environmental protection", "environmental compliance"]
            }

            keywords = reg_keywords.get(regulation, [regulation.lower()])
            found = any(keyword in text_lower for keyword in keywords)

            compliance_status[regulation] = {
                "mentioned": found,
                "status": "addressed" if found else "needs_attention",
                "keywords_found": [k for k in keywords if k in text_lower]
            }

        return {
            "regulations_analysis": compliance_status,
            "compliance_score": len([r for r in compliance_status.values() if r["mentioned"]]) / len(regulations) * 10 if regulations else 10,
            "gaps": [reg for reg, status in compliance_status.items() if not status["mentioned"]]
        }
```

```python
# ========================================
# CELL 7: FILE UPLOAD AND PROCESSING
# ========================================

def upload_and_analyze_contract():
    """Upload and analyze contract with enhanced processing"""

    print("📤 Please upload your contract file")
    print("Supported formats: PDF, DOCX, TXT, PNG, JPG, JPEG, TIFF, BMP")
    print("Note: Scanned PDFs and images will be processed with OCR")

    uploaded = files.upload()

    if not uploaded:
        print("❌ No file uploaded.")
        return None

    # Get the first uploaded file
    filename = next(iter(uploaded))
    file_content = uploaded[filename]

    print(f"\n🔄 Processing {filename} with enhanced capabilities...")

    # Process file with OCR support
    doc_result = analyzer.document_processor.process_uploaded_file(file_content, filename)

    if doc_result["processing_status"] == "error":
        print(f"❌ Error: {doc_result['text']}")
        return None

    print(f"✅ Document processed successfully!")
    print(f"   Format: {doc_result['format'].upper()}")
    print(f"   Words: {doc_result['word_count']:,}")
    print(f"   Characters: {doc_result['char_count']:,}")

    # Get industry selection
    print("\n🏭 Select industry for specialized analysis:")
    industries = ["technology", "construction", "supply_chain", "garment", "general"]

    for i, industry in enumerate(industries, 1):
        print(f"{i}. {industry.title()}")

    while True:
        try:
            choice = input(f"Enter choice (1-{len(industries)}): ").strip()
            industry_index = int(choice) - 1
            if 0 <= industry_index < len(industries):
                selected_industry = industries[industry_index]
                break
            else:
                print("Invalid choice. Please try again.")
        except ValueError:
            print("Please enter a valid number.")

    print(f"\n🔍 Analyzing as {selected_industry} contract using ChainSight methodology...")

    # Perform enhanced analysis
    analysis_result = analyzer.analyze_contract(doc_result["text"], selected_industry)

    # Add document metadata
    analysis_result["document_metadata"] = doc_result

    return analysis_result

# Run the upload and analysis
contract_analysis = upload_and_analyze_contract()
contract_analysis
```

```python
# ========================================
# CELL 8: DISPLAY ENHANCED RESULTS
# ========================================

def display_enhanced_analysis(result):
    """Display comprehensive analysis results"""

    if not result or "error" in result:
        print(f"❌ Analysis failed: {result.get('error', 'Unknown error')}")
        return

    print("\n" + "="*80)
    print("📊 CHAINSIGHT CONTRACT INTELLIGENCE ANALYSIS REPORT")
    print("="*80)

    # Document Information
    doc_meta = result.get("document_metadata", {})
    contract_info = result.get("contract_info", {})

    print(f"\n📋 Document Information:")
    print(f"   File: {doc_meta.get('filename', 'Unknown')}")
    print(f"   Format: {doc_meta.get('format', 'Unknown').upper()}")
    print(f"   Industry: {contract_info.get('industry', 'Unknown').title()}")
    print(f"   Words: {contract_info.get('word_count', 0):,}")
    print(f"   Analysis Method: {contract_info.get('analysis_method', 'Standard')}")
    print(f"   Date: {contract_info.get('analysis_date', 'Unknown')}")

    # Executive Summary
    exec_summary = result.get("executive_summary", {})
    print(f"\n🎯 Executive Summary:")
    print(f"   Overall Risk Score: {exec_summary.get('overall_risk_score', 0):.1f}/10")
    print(f"   Contract Completeness: {exec_summary.get('completeness_score', 0):.1f}/10")
    print(f"   Critical Issues: {exec_summary.get('critical_issues', 0)}")
    print(f"   Priority Level: {exec_summary.get('priority_level', 'Unknown')}")

    # Risk Assessment
    risk_assessment = result.get("risk_assessment", {})
    risk_breakdown = risk_assessment.get("risk_breakdown", {})

    print(f"\n⚠️ Enhanced Risk Assessment:")
    for risk_type, details in risk_breakdown.items():
        severity = details.get("severity", "Unknown")
        score = details.get("score", 0)
        industry_relevant = details.get("industry_relevant", False)

        emoji = "🔴" if severity == "High" else "🟡" if severity == "Medium" else "🟢"
        relevance = " 🎯" if industry_relevant else ""

        print(f"   {emoji} {risk_type.replace('_', ' ').title()}: {score}/10 ({severity}){relevance}")

        if details.get("found_keywords"):
            print(f"      Keywords: {', '.join(details['found_keywords'][:3])}")

    high_risk_areas = risk_assessment.get("high_risk_areas", [])
    if high_risk_areas:
        print(f"\n   🚨 High Risk Areas: {', '.join([area.replace('_', ' ').title() for area in high_risk_areas])}")

    # Missing Clauses
    missing_clauses = result.get("missing_clauses", [])
    if missing_clauses:
        critical_missing = [c for c in missing_clauses if c.get("importance") == "critical"]
        recommended_missing = [c for c in missing_clauses if c.get("importance") == "recommended"]

        print(f"\n📝 Missing Clauses Analysis:")

        if critical_missing:
            print(f"   🚨 Critical Missing ({len(critical_missing)}):")
            for clause in critical_missing[:5]:
                print(f"      - {clause['clause']}")
                print(f"        Reason: {clause['reason']}")

        if recommended_missing:
            print(f"   💡 Recommended Additional ({len(recommended_missing)}):")
            for clause in recommended_missing[:3]:
                print(f"      - {clause['clause']}")
    else:
        print(f"\n✅ No critical clauses missing")

    # AI Analysis
    ai_analysis = result.get("ai_analysis", "")
    if ai_analysis and not ai_analysis.startswith("ChainSight AI analysis unavailable"):
        print(f"\n🤖 ChainSight AI Expert Analysis:")
        print(f"   {ai_analysis}")

    # Compliance Analysis
    compliance = result.get("compliance_analysis", {})
    if compliance:
        compliance_score = compliance.get("compliance_score", 0)
        gaps = compliance.get("gaps", [])

        print(f"\n📋 Regulatory Compliance:")
        print(f"   Compliance Score: {compliance_score:.1f}/10")

        if gaps:
            print(f"   Gaps: {', '.join(gaps)}")
        else:
            print(f"   ✅ All relevant regulations addressed")

    # ChainSight Recommendations
    recommendations = result.get("recommendations", [])
    if recommendations:
        print(f"\n🎯 ChainSight Strategic Recommendations:")

        for rec in recommendations:
            phase = rec.get("phase", "Unknown")
            priority = rec.get("priority", "Medium")
            category = rec.get("category", "General")
            description = rec.get("description", "")
            timeline = rec.get("timeline", "TBD")

            priority_emoji = "🚨" if priority == "HIGH" else "⚠️" if priority == "MEDIUM" else "💡"

            print(f"   {priority_emoji} {phase} Phase - {category}")
            print(f"      Description: {description}")
            print(f"      Timeline: {timeline}")
            print()

# Display the analysis results
if contract_analysis:
    display_enhanced_analysis(contract_analysis)
```
